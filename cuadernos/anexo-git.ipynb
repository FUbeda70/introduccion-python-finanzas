# 1. Crear una cuenta en GitHub

Abre:

**https://github.com/**

Pulsa **Sign up** y crea una cuenta con:

- correo electrónico;
- nombre de usuario;
- contraseña.

Después verifica tu correo.

> ✅ Recomendación: utiliza un nombre de usuario profesional, especialmente si vas a usar GitHub para trabajos académicos o profesionales.

# 2. Instalar Git en Windows

Git es el programa que permite guardar versiones de tus proyectos y sincronizarlos con GitHub.

Abre:

**https://git-scm.com/**

Descarga Git para Windows e instálalo. Si no sabes qué opción elegir durante la instalación, puedes aceptar las opciones por defecto.

## Comprobar la instalación

Abre `cmd` y escribe:

```bash
git --version
```

Deberías ver algo parecido a:

```text
git version 2.x.x
```

# 3. Configurar Git por primera vez

Git necesita saber qué nombre y correo debe asociar a tus cambios.

```bash
git config --global user.name "Tu Nombre"
git config --global user.email "tu_email@ejemplo.com"
```

Ejemplo:

```bash
git config --global user.name "Ana García"
git config --global user.email "ana.garcia@estudiante.uam.es"
```

Comprueba la configuración:

```bash
git config --global --list
```

# 4. Crear un proyecto de ejemplo

Crea una carpeta:

```text
Documentos/
└── mi_primer_repositorio/
```

Dentro crea un archivo:

```text
README.md
```

con este contenido:

```markdown
# Mi primer repositorio

Este es mi primer proyecto usando Git y GitHub.
```

````python
print("# Mi primer repositorio")
print()
print("Este es mi primer proyecto usando Git y GitHub.")
````

# 5. Convertir la carpeta en un repositorio Git

Abre una terminal dentro de la carpeta del proyecto y ejecuta:

```bash
git init
```

Ahora comprueba el estado:

```bash
git status
```

Git te mostrará los archivos que todavía no están siendo controlados.

# 6. Crear el primer commit

Añade todos los archivos:

```bash
git add .
```

Crea el primer commit:

```bash
git commit -m "Primer commit"
```

## 🧠 Idea clave

Un **commit** es una fotografía del estado del proyecto en un momento determinado.

# 7. Crear un repositorio en GitHub

En GitHub:

1. Pulsa **New repository**.
2. Nombre: `mi_primer_repositorio`.
3. Selecciona **Public** o **Private**.
4. Si ya tienes `README.md` localmente, no marques la opción de crear otro README.
5. Pulsa **Create repository**.

GitHub mostrará una URL parecida a:

```text
https://github.com/TU_USUARIO/mi_primer_repositorio.git
```

# 8. Conectar el repositorio local con GitHub

En la terminal:

```bash
git remote add origin https://github.com/TU_USUARIO/mi_primer_repositorio.git
git branch -M main
git push -u origin main
```

La primera vez, Git puede abrir el navegador para que inicies sesión y autorices la conexión.

# 9. Flujo habitual con comandos Git

Cada vez que modifiques el proyecto:

### 1. Ver cambios

```bash
git status
```

### 2. Añadir cambios

```bash
git add .
```

### 3. Crear commit

```bash
git commit -m "Descripción breve del cambio"
```

### 4. Subir a GitHub

```bash
git push
```

## 🔁 Flujo que debes recordar

```text
Modificar archivos
      ↓
git add .
      ↓
git commit -m "mensaje"
      ↓
git push
```

# 11. Descargar cambios desde GitHub

Si el repositorio ha cambiado en GitHub:

```bash
git pull
```

Una rutina habitual es:

```text
git pull
↓
trabajar
↓
git add .
↓
git commit
↓
git push
```

# 12. Clonar un repositorio existente

Si el proyecto ya existe en GitHub y quieres descargarlo:

```bash
git clone https://github.com/TU_USUARIO/mi_primer_repositorio.git
```

Después entra en la carpeta:

```bash
cd mi_primer_repositorio
```

# 13. Trabajar con Git desde VS Code

Abre la carpeta del proyecto:

**File → Open Folder**

En la barra lateral izquierda encontrarás:

### 🌿 Source Control

Desde ahí puedes:

- ver archivos modificados;
- preparar cambios;
- hacer commits;
- hacer push;
- hacer pull.

# 14. Conectar VS Code con GitHub

En VS Code:

1. Pulsa el icono de cuenta.
2. Selecciona **Sign in with GitHub**.
3. Se abrirá el navegador.
4. Autoriza Visual Studio Code.
5. Regresa a VS Code.

Tu cuenta quedará conectada.

# 15. Hacer un commit desde VS Code

Después de modificar un archivo:

1. Abre **Source Control**.
2. Pulsa `+` junto al archivo para añadirlo al staging area.
3. Escribe un mensaje de commit.
4. Pulsa **Commit**.
5. Pulsa **Sync Changes** o **Push**.

Ejemplo de mensaje:

```text
Actualiza descripción del proyecto
```

# 16. Terminal vs VS Code

| Acción | Terminal | VS Code |
|---|---|---|
| Ver cambios | `git status` | Source Control |
| Añadir | `git add .` | Stage Changes |
| Commit | `git commit -m "..."` | Commit |
| Subir | `git push` | Push / Sync Changes |
| Descargar | `git pull` | Pull / Sync Changes |

> 💡 Es recomendable conocer los comandos básicos aunque después uses principalmente VS Code.

# 17. Instalar GitHub Desktop

Abre:

**https://desktop.github.com/**

Descarga e instala **GitHub Desktop**.

Después abre la aplicación.

# 18. Iniciar sesión en GitHub Desktop

1. Selecciona **Sign in to GitHub.com**.
2. Se abrirá el navegador.
3. Autoriza GitHub Desktop.
4. Regresa a la aplicación.

Tu cuenta quedará conectada.

# 19. Clonar un repositorio con GitHub Desktop

En GitHub Desktop:

**File → Clone repository**

Selecciona tu repositorio y elige dónde guardarlo.

Pulsa:

**Clone**

GitHub Desktop descargará automáticamente el proyecto.

# 20. Actualizar GitHub con GitHub Desktop

Después de modificar un archivo:

1. Abre GitHub Desktop.
2. Verás los archivos modificados.
3. Escribe un mensaje de commit.
4. Pulsa **Commit to main**.
5. Pulsa **Push origin**.

```text
Modificar
↓
Commit to main
↓
Push origin
```

# 21. Descargar cambios con GitHub Desktop

Pulsa:

**Fetch origin**

Si hay cambios nuevos aparecerá:

**Pull origin**

Pulsa **Pull origin** para descargarlos.

# 22. ¿Qué opción conviene utilizar?

### 🖥️ Terminal
- universal;
- aprendes cómo funciona Git;
- útil en cualquier entorno.

### 🧑‍💻 VS Code
- integrado con tu código;
- cómodo para programar;
- permite combinar interfaz gráfica y terminal.

### 🐙 GitHub Desktop
- muy sencillo para principiantes;
- permite visualizar los cambios;
- evita memorizar muchos comandos.

# 23. Conceptos fundamentales

| Concepto | Significado |
|---|---|
| Repository | Carpeta de proyecto gestionada por Git |
| Commit | Fotografía de los cambios |
| Push | Envía commits a GitHub |
| Pull | Descarga cambios desde GitHub |
| Clone | Descarga un repositorio por primera vez |
| Branch | Versión paralela del proyecto |
| Remote | Repositorio remoto, normalmente GitHub |

# 25. Problemas frecuentes

## Git no se reconoce como comando

Reinicia la terminal o VS Code y comprueba:

```bash
git --version
```

## `fatal: not a git repository`

No estás dentro de la carpeta correcta.

```bash
cd ruta_de_tu_proyecto
```

## GitHub tiene cambios que no tengo

```bash
git pull
```

## `git push` pide autenticación

Lo normal es que GitHub abra el navegador para que autorices la operación.
