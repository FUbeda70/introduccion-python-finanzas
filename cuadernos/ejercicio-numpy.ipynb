# Las acciones siguen un proceso estocástico

- Un proceso estocástico es una secuencia de variables aleatorios.
  - Simular un proceso estocásticos es generar una secuencia trayectorias aleatorias.
  - **Propiedad de Markov**: los valores de mañana del activo sólo dependen del estado de hoy del proceso. Los otros valores del pasado no le afectan. Es un proceso desmemoriado.

  El comportamiento de una acción puede definirse como un **proceso Browniano Geométrico**, el cual viene definido por la siguiente ecuación estocástica diferencial:
  $$dS_t=\mu S_tdt+\sigma S_t dZ_t$$

  $Z$ es un proceso Browiano y sus propiedades básicas son:

  $$Z_0=0$$

  y su incrementos cumplen:
  $$Z_{t+\Delta t}-Z_t\sim N(0,\Delta t)$$

  Durante el intervalo $\Delta t$ el camino aleatorio viene definido por:
  $$E[\Delta Z]=0$$
  $$Var[\Delta Z]=\Delta t$$

  Por ello se describe:

  $$\Delta Z =\sqrt{\Delta t} \varepsilon$$

  donde:
  $$\varepsilon\sim N(0,1)$$

Si retomamos el comportamiento de la accción:
$$\frac{dS_t}{S_t}=\mu dt+\sigma dZ_t$$

- Componente determinista o tendencia: $\mu dt$
- Componente aleatorio: $\sigma dZ_t$
- Shock browiano: $dZ_t$

La ecuación está en tiempo continuo y no podemos simular infinitos instantes en Python necesitamos discretizar la expresión.

### Esquema de Euler-Maruyama

Partimos de:

$$dS_t=\mu S_tdt+\sigma S_t dZ_t$$

El esquema de Euler sustituye los diferenciales por incrementos finitos:
$$dS_t \approx S_{t+\Delta t}-S_t$$
$$dt \approx \Delta t$$

y apra el browniano:

$$dZ_t \approx \sqrt{\Delta t} \varepsilon_t$$

donde:

$$\varepsilon_t=N(0,1)$$

Sustituyendo:

$$S_{t+\Delta t}-S_t=\mu S_t \Delta t+ \sigma S_t\sqrt{\Delta t} \varepsilon_t$$


$$S_{t+\Delta t}=S_t(1+\mu \Delta t+ \sigma S_t\sqrt{\Delta t} \varepsilon_t)$$
