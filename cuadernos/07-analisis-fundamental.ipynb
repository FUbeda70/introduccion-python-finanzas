# ROE

## Rentabilidad Financiera
$$ROE=\frac{\text{Beneficio Neto}}{Patrimonio}$$

## Apalancamiento Financiero

$$ROC=\frac{\text{Beneficio sin Intereses y Después de Impuestos}}{Activo Total}$$

$$ROE=ROC+\left(ROC-k_d(1-\tau)\times \frac{Deuda}{Patrimonio}\right)$$

$$k_d=\frac{Intereses}{Deuda}$$


$$\tau=\frac{Impuestos}{Beneficio Neto}$$

# RATIOS DE MERCADO

$$Price-to-book=\frac{\text{Capitalización Bursatil}}{\text{Patrimonio Neto}}$$


$$PER=\frac{Precio}{\text{Benefcio Neto}}$$



```python
empresa = yf.Ticker("ITX.MC")

price_to_book = empresa.info.get("priceToBook")
per = empresa.info.get("trailingPE")

print(price_to_book,per)
```
