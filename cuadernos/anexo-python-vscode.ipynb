> 💡 No necesitas conocimientos previos de programación.

# 1. Instalar Python

## Paso 1.1 — Descargar Python

Abre tu navegador y entra en:

**https://www.python.org/downloads/**

Pulsa el botón de descarga de la versión estable más reciente de Python 3.

## Paso 1.2 — Ejecutar el instalador

Cuando se abra el instalador:

> ⚠️ **MUY IMPORTANTE:** antes de pulsar **Install Now**, marca **Add python.exe to PATH**.

Después pulsa **Install Now**.

# 2. Comprobar que Python está instalado

Abre una terminal de Windows:

- Pulsa `Windows`.
- Escribe `cmd`.
- Abre **Símbolo del sistema**.

Escribe:

```bash
python --version
```

Deberías ver algo parecido a:

```text
Python 3.13.x
```

Comprueba también `pip`:

```bash
pip --version
```

`pip` sirve para instalar librerías de Python.

````python
import sys
print("Versión de Python:")
print(sys.version)
````

# 3. Instalar Visual Studio Code

Abre:

**https://code.visualstudio.com/**

Descarga la versión para **Windows**.

Durante la instalación puedes aceptar las opciones por defecto.

Recomendado:

- Crear acceso directo.
- Añadir VS Code al PATH.
- Añadir **Open with Code** al menú contextual.

# 4. Instalar las extensiones necesarias

En VS Code, pulsa el icono **Extensions** de la barra lateral izquierda.

Busca e instala:

### ✅ Python
Publicada por **Microsoft**

### ✅ Jupyter
Publicada por **Microsoft**

**Python** permite ejecutar archivos `.py`.

**Jupyter** permite abrir y ejecutar notebooks `.ipynb`.

# 5. Crear una carpeta de trabajo

Crea una carpeta, por ejemplo:

```text
Documentos/
└── PythonFinance/
```

Después, en VS Code:

**File → Open Folder**

Selecciona `PythonFinance`.

> 💡 Es recomendable trabajar siempre dentro de una carpeta de proyecto.

# 6. Abrir la terminal de VS Code

En VS Code:

**Terminal → New Terminal**

Comprueba que Python funciona:

```bash
python --version
```

# 7. Crear un entorno virtual

En la terminal escribe:

```bash
python -m venv .venv
```

Se creará una carpeta llamada:

```text
.venv
```

Un entorno virtual permite instalar librerías para un proyecto sin afectar al resto del ordenador.

# 8. Activar el entorno virtual

En Windows escribe:

```bash
.venv\Scripts\activate
```

Si funciona, verás algo parecido a:

```text
(.venv) C:\Users\TuNombre\Documents\PythonFinance>
```

El texto `(.venv)` indica que el entorno virtual está activo.

# 9. Seleccionar el intérprete en VS Code

Pulsa:

`Ctrl + Shift + P`

Escribe:

```text
Python: Select Interpreter
```

Selecciona el intérprete que contiene:

```text
.venv
```

# 10. Instalar librerías básicas

Con el entorno virtual activado:

```bash
pip install numpy pandas matplotlib scipy scikit-learn jupyter
```

| Librería | Uso |
|---|---|
| NumPy | Cálculo numérico |
| Pandas | Datos y tablas |
| Matplotlib | Gráficos |
| SciPy | Estadística y matemáticas |
| Scikit-learn | Machine Learning |
| Jupyter | Notebooks |

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Todo funciona correctamente ✅")
````

# 11. Crear tu primer programa Python

En VS Code crea un archivo:

```text
hola.py
```

Escribe:

```python
print("Hola, Python")
```

Guárdalo y ejecuta:

```bash
python hola.py
```

Deberías ver:

```text
Hola, Python
```

````python
print("Hola, Python 👋")
````

# 12. Crear tu primer notebook

En VS Code:

**File → New File → Jupyter Notebook**

o crea:

```text
primer_notebook.ipynb
```

Un notebook tiene dos tipos principales de celdas:

### 📝 Markdown
Para texto, explicaciones y fórmulas.

### 🐍 Code
Para ejecutar código Python.

````python
x = 10
y = 20
x + y
````

# 13. Seleccionar el kernel

En la esquina superior derecha del notebook aparecerá **Select Kernel**.

Selecciona:

```text
.venv
```

Si no aparece:

1. Pulsa **Select Kernel**.
2. Selecciona **Python Environments**.
3. Elige `.venv`.

> ⚠️ El kernel determina qué instalación de Python ejecuta el notebook.

# 14. Prueba final

Ejecuta la siguiente celda.

Si aparece un gráfico, la instalación está lista.

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.title("Mi primer gráfico con Python")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
````

# 🆘 Problemas frecuentes

## `python` no se reconoce como comando

Vuelve a ejecutar el instalador de Python y asegúrate de activar la opción de añadir Python al `PATH`.

## VS Code no encuentra Python

Pulsa `Ctrl + Shift + P` → **Python: Select Interpreter** → selecciona `.venv`.

## Falta una librería

Con `.venv` activado:

```bash
pip install nombre_libreria
```

Ejemplo:

```bash
pip install pandas
```

## El notebook usa otro Python

Pulsa **Select Kernel** y selecciona `.venv`.
